# Named Entity Recognition (NER) using BERT
We will use a pre-trained BERT model (`dbmdz/bert-large-cased-finetuned-conll03-english`) from Hugging Face for token classification.

```python
!pip install -q transformers
```

```python
from transformers import pipeline

# Initialize the NER pipeline with the specific BERT model
bert_ner = pipeline(
    "ner",
    model="dbmdz/bert-large-cased-finetuned-conll03-english",
    aggregation_strategy="simple"
)
```

```python
sample_text = "Google, headquartered in Mountain View, California, was founded by Larry Page and Sergey Brin."

results = bert_ner(sample_text)
for entity in results:
    print(f"Entity: {entity['word']:<15} | Type: {entity['entity_group']:<5} | Score: {entity['score']:.4f}")
```
