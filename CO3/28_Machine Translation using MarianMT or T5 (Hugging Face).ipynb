# Machine Translation with Hugging Face (MarianMT & T5)

This notebook demonstrates how to perform machine translation using Hugging Face's `transformers` library. We will explore two approaches:
1. **MarianMT**: Ideal for specific, high-quality translation pairs (e.g., English to Spanish).
2. **T5 (Text-to-Text Transfer Transformer)**: A versatile model capable of translation among other NLP tasks.

## 1. Translation using MarianMT

MarianMT models are trained on the OPUS dataset and are excellent for specific translation directions.

```python
from transformers import MarianMTModel, MarianTokenizer

# Using English to Romance (covers French, Spanish, Portuguese, etc.) model, or a specific language pair like English to French
model_name = 'Helsinki-NLP/opus-mt-en-fr'
tokenizer = MarianTokenizer.from_pretrained(model_name)
model = MarianMTModel.from_pretrained(model_name)

def translate_marian(text):
    # Tokenize input text
    inputs = tokenizer(text, return_tensors='pt', padding=True)
    # Generate translation output
    translated = model.generate(**inputs)
    # Decode the output
    translated_text = [tokenizer.decode(t, skip_special_tokens=True) for t in translated]
    return translated_text[0]

# Test MarianMT translation
english_text = "Hello, how are you doing today? I am learning machine translation."
french_translation = translate_marian(english_text)
print(f"Source (English): {english_text}")
print(f"Translation (French): {french_translation}")
```

## 2. Translation using T5

T5 uses a prefix-based instruction format (e.g., `translate English to German: ...`) to determine the task.

```python
from transformers import T5ForConditionalGeneration, T5Tokenizer

# Loading a standard t5-small model
t5_model_name = 't5-small'
t5_tokenizer = T5Tokenizer.from_pretrained(t5_model_name, legacy=False)
t5_model = T5ForConditionalGeneration.from_pretrained(t5_model_name)

def translate_t5(text, source_lang="English", target_lang="German"):
    # Prefix required by T5
    input_text = f"translate {source_lang} to {target_lang}: {text}"
    inputs = t5_tokenizer(input_text, return_tensors='pt')
    # Generate output
    outputs = t5_model.generate(**inputs, max_length=40)
    # Decode output
    translated_text = t5_tokenizer.decode(outputs[0], skip_special_tokens=True)
    return translated_text

# Test T5 translation
input_phrase = "Machine learning is transforming our world."
german_translation = translate_t5(input_phrase, "English", "German")
print(f"Source (English): {input_phrase}")
print(f"Translation (German): {german_translation}")
```
