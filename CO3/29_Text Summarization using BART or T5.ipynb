# Text Summarization with BART

This notebook demonstrates how to perform text summarization using a pre-trained BART (`facebook/bart-large-cnn`) model from the Hugging Face Transformers library.

```python
!pip install -q transformers torch
```

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

# Define the pre-trained BART model
model_name = "facebook/bart-large-cnn"

# Load the tokenizer and model explicitly
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)
```

```python
# Sample text to summarize
article = """
New York (CNN) -- When the safe-deposit box was opened at a bank in upstate New York, it contained a treasure trove of historical documents, including letters written by George Washington, Thomas Jefferson and Alexander Hamilton.

The letters, which had been stored in the box for at least 70 years, were found at the KeyBank branch in Albany by state archivists who were searching for unclaimed property.

State officials said the documents are in excellent condition and will be transferred to the New York State Archives, where they will be preserved and eventually made available to the public. Among the highlights is a 1779 letter from Washington to Gen. James Clinton, discussing military strategy during the Revolutionary War.
"""

# Tokenize and generate summary
inputs = tokenizer([article], max_length=1024, return_tensors="pt", truncation=True)
summary_ids = model.generate(inputs["input_ids"], num_beams=4, min_length=30, max_length=130, early_stopping=True)

# Decode and display summary
summary = tokenizer.decode(summary_ids[0], skip_special_tokens=True)
print("Original Text:\n", article)
print("\nSummarized Text:\n", summary)
```
