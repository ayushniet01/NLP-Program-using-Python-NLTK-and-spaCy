# Text Similarity using Word Mover's Distance (WMD)

```python
import gensim.downloader as api
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
import numpy as np

print("Loading Word2Vec model (this may take a few minutes)...")
model = api.load('word2vec-google-news-300')
print("Model loaded successfully.")
```

## Preprocessing and Calculating WMD



```python
# Preprocessing function
def preprocess(sentence):
    stop_words = set(stopwords.words('english'))
    words = word_tokenize(sentence.lower())
    # Return only alphabetic tokens that are not stopwords
    return [word for word in words if word.isalpha() and word not in stop_words]

# Sample sentences to compare
sentence_1 = "Obamacare affects the healthcare of millions of Americans."
sentence_2 = "The modern health insurance reform impacts many US citizens."
sentence_3 = "An orange is a delicious citrus fruit."

# Preprocess sentences
words_1 = preprocess(sentence_1)
words_2 = preprocess(sentence_2)
words_3 = preprocess(sentence_3)

print("Processed Sentence 1:", words_1)
print("Processed Sentence 2:", words_2)
print("Processed Sentence 3:", words_3)
```

```python
# Calculate Word Mover's Distance (WMD)
# Note: A smaller distance indicates higher semantic similarity.

distance_1_2 = model.wmdistance(words_1, words_2)
distance_1_3 = model.wmdistance(words_1, words_3)

print(f"WMD between Sentence 1 and Sentence 2 (Similar topic): {distance_1_2:.4f}")
print(f"WMD between Sentence 1 and Sentence 3 (Different topic): {distance_1_3:.4f}")
```
