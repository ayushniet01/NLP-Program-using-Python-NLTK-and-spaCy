# Sentiment Analysis using TextBlob and VADER

```python
# Install and import necessary libraries
!pip install textblob nltk

import nltk
nltk.download('vader_lexicon')

from textblob import TextBlob
from nltk.sentiment.vader import SentimentIntensityAnalyzer
```

```python
# Define sample text for sentiment analysis
sample_texts = [
    "I absolutely love this product! It has changed my life for the better.",
    "This is the worst service I have ever experienced. Completely unacceptable.",
    "The package arrived today. It is what it is."
]

# Initialize VADER sentiment analyzer
sia = SentimentIntensityAnalyzer()

print("--- Sentiment Analysis Comparison ---\n")
for text in sample_texts:
    # TextBlob analysis
    blob = TextBlob(text)
    tb_polarity = blob.sentiment.polarity

    # VADER analysis
    vader_scores = sia.polarity_scores(text)
    vader_compound = vader_scores['compound']

    print(f"Text: '{text}'")
    print(f"  TextBlob Polarity: {tb_polarity:.4f}")
    print(f"  VADER Compound Score: {vader_compound:.4f}")
    print("-" * 40)
```

### Comparison of TextBlob and VADER

| Feature | TextBlob | VADER |
| :--- | :--- | :--- |
| **Methodology** | Lexicon-based | Lexicon and Rule-based (optimized for social media) |
| **Key Scores** | Polarity [-1, 1], Subjectivity [0, 1] | Positive, Neutral, Negative, and Compound [-1, 1] |
| **Slang & Emojis** | Limited capability | Strong native support (e.g., "LOL", ":)", "great!!!") |
| **Best For** | General and formal text | Informal text, tweets, reviews, and short-form web content |
