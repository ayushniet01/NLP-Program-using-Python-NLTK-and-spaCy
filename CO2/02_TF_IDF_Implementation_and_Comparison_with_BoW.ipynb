```python
corpus = [
    'The quick brown fox jumps over the lazy dog',
    'Never jump over the lazy dog again',
    'The quick brown fox is quick and brown'
]

print("Our sample corpus:")
for i, doc in enumerate(corpus):
    print(f"Document {i+1}: {doc}")
```

```python
import nltk
from nltk.tokenize import word_tokenize
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)


preprocessed_corpus = []
for doc in corpus:

    tokens = word_tokenize(doc.lower())
    preprocessed_corpus.append(tokens)

print("\nPreprocessed Corpus:")
for i, doc_tokens in enumerate(preprocessed_corpus):
    print(f"Document {i+1} tokens: {doc_tokens}")
```

```python
from collections import Counter

all_words = []
for doc_tokens in preprocessed_corpus:
    all_words.extend(doc_tokens)

vocabulary = sorted(list(set(all_words)))
print(f"\nVocabulary ({len(vocabulary)} words): {vocabulary}")

bow_vectors = []
for doc_tokens in preprocessed_corpus:

    word_counts = Counter(doc_tokens)

    bow_vector = [word_counts[word] for word in vocabulary]
    bow_vectors.append(bow_vector)

print("\nBag-of-Words Vectors:")
for i, vector in enumerate(bow_vectors):
    print(f"Document {i+1} BoW: {vector}")
```

```python
import math

def calculate_tf(document_tokens, word):
    return document_tokens.count(word) / len(document_tokens)

tf_scores = []
for doc_tokens in preprocessed_corpus:
    doc_tf = {}
    for word in vocabulary:
        tf = calculate_tf(doc_tokens, word)
        doc_tf[word] = tf
    tf_scores.append(doc_tf)

print("\nTerm Frequency (TF) Scores:")
for i, tf_doc in enumerate(tf_scores):
    print(f"Document {i+1} TF: {tf_doc}")
```

```python

def calculate_idf(word, corpus_tokens):
    document_count = len(corpus_tokens)

    docs_with_word = sum(1 for doc in corpus_tokens if word in doc)

    return math.log(document_count / (docs_with_word + 1))

idf_scores = {}
for word in vocabulary:
    idf_scores[word] = calculate_idf(word, preprocessed_corpus)

print("\nInverse Document Frequency (IDF) Scores:")
print(idf_scores)
```

```python

tfidf_vectors = []
for i, doc_tf in enumerate(tf_scores):
    tfidf_doc_vector = []
    for word in vocabulary:
        tfidf = doc_tf[word] * idf_scores[word]
        tfidf_doc_vector.append(tfidf)
    tfidf_vectors.append(tfidf_doc_vector)

print("\nTF-IDF Vectors:")
for i, vector in enumerate(tfidf_vectors):
    print(f"Document {i+1} TF-IDF: {vector}")
```
