### 1. Import Libraries and Load Dataset


```python
import numpy as np
import pandas as pd
from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.svm import LinearSVC
from sklearn.pipeline import Pipeline
from sklearn.metrics import classification_report, accuracy_score, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

# Load a subset of categories for classification
categories = ['alt.atheism', 'soc.religion.christian', 'comp.graphics', 'sci.med']

train_data = fetch_20newsgroups(subset='train', categories=categories, shuffle=True, random_state=42)
test_data = fetch_20newsgroups(subset='test', categories=categories, shuffle=True, random_state=42)

print(f"Training samples: {len(train_data.data)}")
print(f"Testing samples: {len(test_data.data)}")
```

### 2. Naïve Bayes Pipeline with TF-IDF


```python
# Create a pipeline combining TF-IDF and Naïve Bayes
nb_pipeline = Pipeline([
    ('tfidf', TfidfVectorizer(stop_words='english')),
    ('clf', MultinomialNB())
])

# Train the model
nb_pipeline.fit(train_data.data, train_data.target)

# Predict on test data
nb_predictions = nb_pipeline.predict(test_data.data)

# Evaluate
nb_accuracy = accuracy_score(test_data.target, nb_predictions)
print(f"Naïve Bayes Accuracy: {nb_accuracy:.4f}\n")
print(classification_report(test_data.target, nb_predictions, target_names=test_data.target_names))
```

### 3. Support Vector Machine (SVM) Pipeline with TF-IDF


```python
# Create a pipeline combining TF-IDF and Linear SVM
svm_pipeline = Pipeline([
    ('tfidf', TfidfVectorizer(stop_words='english')),
    ('clf', LinearSVC(random_state=42))
])

# Train the model
svm_pipeline.fit(train_data.data, train_data.target)

# Predict on test data
svm_predictions = svm_pipeline.predict(test_data.data)

# Evaluate
svm_accuracy = accuracy_score(test_data.target, svm_predictions)
print(f"SVM Accuracy: {svm_accuracy:.4f}\n")
print(classification_report(test_data.target, svm_predictions, target_names=test_data.target_names))
```

### 4. Compare Model Performance and Visualize


```python
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Plot Naïve Bayes Confusion Matrix
ConfusionMatrixDisplay.from_predictions(
    test_data.target, nb_predictions,
    display_labels=test_data.target_names,
    ax=axes[0], cmap='Blues', xticks_rotation=45
)
axes[0].set_title(f"Naïve Bayes (Acc: {nb_accuracy:.2%})")

# Plot SVM Confusion Matrix
ConfusionMatrixDisplay.from_predictions(
    test_data.target, svm_predictions,
    display_labels=test_data.target_names,
    ax=axes[1], cmap='Greens', xticks_rotation=45
)
axes[1].set_title(f"Linear SVM (Acc: {svm_accuracy:.2%})")

plt.tight_layout()
plt.show()
```
