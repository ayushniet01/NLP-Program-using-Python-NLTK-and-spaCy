### Topic Modeling using Latent Semantic Analysis (LSA)



```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD

# 1. Create a sample dataset of documents representing different topics
documents = [
    "The quick brown fox jumps over the lazy dog.",
    "Artificial Intelligence and Machine Learning are transforming industries.",
    "Deep learning neural networks require massive computational power and GPUs.",
    "A healthy diet and regular cardiovascular exercise improve physical health.",
    "Proper nutrition, hydration, and sleep are essential for fitness training.",
    "Natural Language Processing helps computers understand and interpret human language.",
    "Robots and automation are being integrated into manufacturing lines.",
    "Yoga, meditation, and mental health awareness reduce daily stress."
]

df = pd.DataFrame({'Document': documents})
display(df)
```

### Step 2: TF-IDF Vectorization


```python
# Vectorize the text using TF-IDF
vectorizer = TfidfVectorizer(stop_words='english')
tfidf_matrix = vectorizer.fit_transform(df['Document'])

print(f"TF-IDF Matrix Shape: {tfidf_matrix.shape}")
```

### Step 3: Dimensionality Reduction using Truncated SVD (LSA)


```python
# Define the number of topics to extract
n_topics = 3

svd_model = TruncatedSVD(n_components=n_topics, random_state=42)
svd_matrix = svd_model.fit_transform(tfidf_matrix)

print(f"SVD Matrix Shape: {svd_matrix.shape}")
```

### Step 4: Extract and Display Key Words per Topic

```python
terms = vectorizer.get_feature_names_out()

for index, component in enumerate(svd_model.components_):
    # Get the terms with the highest weights for this component
    zipped_chunks = zip(terms, component)
    sorted_terms = sorted(zipped_chunks, key=lambda x: x[1], reverse=True)
    top_terms = sorted_terms[:5]

    print(f"\nTopic {index + 1}:")
    for term, weight in top_terms:
        print(f"  - {term}: {weight:.4f}")
```
