# Topic Modeling using Latent Dirichlet Allocation (LDA)



```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
import gensim
from gensim import corpora

# Download necessary NLTK datasets
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
# Sample dataset of documents representing different topics (Technology, Food, and Space)
documents = [
    "The development of quantum computers is accelerating rapidly with new processors.",
    "Artificial intelligence and machine learning are transforming modern software development.",
    "I love cooking Italian pasta with fresh tomatoes, garlic, and olive oil.",
    "Baking homemade sourdough bread requires patience, flour, water, and yeast.",
    "NASA's James Webb Space Telescope is capturing stunning images of distant galaxies.",
    "The SpaceX Falcon heavy rocket launched successfully into orbit yesterday.",
    "New software updates improve computer performance and security features.",
    "A classic Margherita pizza uses fresh mozzarella, basil, and tomato sauce."
]

# Preprocessing setup
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()

def preprocess(text):
    tokens = word_tokenize(text.lower())
    # Filter out non-alphabetic tokens, stopwords, and short words, then lemmatize
    cleaned_tokens = [
        lemmatizer.lemmatize(token)
        for token in tokens
        if token.isalpha() and token not in stop_words and len(token) > 2
    ]
    return cleaned_tokens

processed_docs = [preprocess(doc) for doc in documents]
print("Sample Preprocessed Document:", processed_docs[0])
```

```python
# Create Dictionary and Corpus required for LDA
dictionary = corpora.Dictionary(processed_docs)
corpus = [dictionary.doc2bow(doc) for doc in processed_docs]

# Train LDA Model (Setting number of topics to 3)
num_topics = 3
lda_model = gensim.models.LdaModel(
    corpus=corpus,
    id2word=dictionary,
    num_topics=num_topics,
    random_state=42,
    passes=15
)

# Display the discovered topics and their keywords
for idx, topic in lda_model.print_topics(-1):
    print(f"Topic {idx}: {topic}\n")
```
