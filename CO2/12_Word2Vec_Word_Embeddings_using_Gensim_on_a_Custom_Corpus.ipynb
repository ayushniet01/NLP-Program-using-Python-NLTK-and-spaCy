# Word2Vec Word Embeddings using Gensim on a Custom Corpus

```python
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
import string

nltk.download('punkt')
nltk.download('punkt_tab')
```

### 1. Define and Preprocess Custom Corpus
Word2Vec expects a list of sentences, where each sentence is represented as a list of lowercased tokens (words).

```python
# A small custom corpus about artificial intelligence, machine learning, and data science
custom_corpus = """
Artificial intelligence and machine learning are transforming the world.
Machine learning models require data to learn patterns and make predictions.
Deep learning is a subset of machine learning inspired by human brain neural networks.
Data science involves collecting, cleaning, analyzing, and interpreting data.
Data scientists use Python, algorithms, and models to extract valuable insights.
Neural networks are crucial for training deep learning computer vision models.
"""

def preprocess_corpus(text):
    # Split the corpus into sentences
    sentences = sent_tokenize(text.strip().lower())

    tokenized_sentences = []
    for sentence in sentences:
        # Tokenize words
        tokens = word_tokenize(sentence)
        # Remove punctuation
        cleaned_tokens = [word for word in tokens if word not in string.punctuation]
        tokenized_sentences.append(cleaned_tokens)

    return tokenized_sentences

processed_sentences = preprocess_corpus(custom_corpus)
print("Sample preprocessed sentences:")
for s in processed_sentences[:3]:
    print(s)
```

### 2. Train Word2Vec Model
We will initialize and train a Continuous Bag of Words (CBOW) Word2Vec model using Gensim.

```python
from gensim.models import Word2Vec

# Hyperparameters
vector_size = 50  # Dimensionality of the word vectors
window = 3        # Maximum distance between the current and predicted word within a sentence
min_count = 1     # Ignores all words with total frequency lower than this
workers = 4       # Use these many worker threads to train the model

# Train the model
model = Word2Vec(
    sentences=processed_sentences,
    vector_size=vector_size,
    window=window,
    min_count=min_count,
    workers=workers,
    sg=0 # sg=0 for CBOW; sg=1 for Skip-gram
)

print("Model training completed.")
```

### 3. Explore the Word Embeddings
Now we can query the model to inspect vocabulary, extract word vectors, and find similarities.

```python
# Get vocabulary
vocab = list(model.wv.key_to_index.keys())
print(f"Vocabulary size: {len(vocab)}")
print("Vocabulary words:", vocab)

# Extract vector for a specific word
word = "learning"
if word in model.wv:
    vector = model.wv[word]
    print(f"\nEmbedding vector for '{word}' (first 10 dimensions):\n", vector[:10])
```

```python
# Find most similar words to a target word
target_word = "data"
if target_word in model.wv:
    similar_words = model.wv.most_similar(target_word, topn=3)
    print(f"Words most similar to '{target_word}':")
    for word, score in similar_words:
        print(f"{word}: {score:.4f}")

# Compute cosine similarity between two words
word1, word2 = "learning", "models"
if word1 in model.wv and word2 in model.wv:
    similarity = model.wv.similarity(word1, word2)
    print(f"\nCosine similarity between '{word1}' and '{word2}': {similarity:.4f}")
```
