### N-Gram Model Generation

An N-gram is a contiguous sequence of n items from a given sample of text or speech.



```python
import collections
import re

corpus = "This is a simple example corpus. This corpus will be used to demonstrate N-gram generation."

print("Original Corpus:")
print(corpus)
```

### 1. Tokenization

First, we need to tokenize the corpus into words. We'll convert text to lowercase and remove punctuation to standardize the words.

```python
tokens = re.findall(r'\b\w+\b', corpus.lower())

print("Tokens:")
print(tokens)
```

### 2. Unigram Generation

Unigrams are simply individual words in the corpus.

```python
def generate_unigrams(tokens):
    return collections.Counter(tokens)

unigrams = generate_unigrams(tokens)
print("Unigrams (word counts):")
for word, count in unigrams.most_common(5):
    print(f"'{word}': {count}")
```

### 3. Bigram Generation

Bigrams are pairs of consecutive words.

```python
def generate_ngrams(tokens, n):
    ngrams = []
    for i in range(len(tokens) - n + 1):
        ngrams.append(tuple(tokens[i : i + n]))
    return collections.Counter(ngrams)

bigrams = generate_ngrams(tokens, 2)
print("Bigrams:")
for bigram, count in bigrams.most_common(5):
    print(f"'{' '.join(bigram)}': {count}")
```

### 4. Trigram Generation

Trigrams are sequences of three consecutive words.

```python
trigrams = generate_ngrams(tokens, 3)
print("Trigrams:")
for trigram, count in trigrams.most_common(5):
    print(f"'{' '.join(trigram)}': {count}")
```
