### Download and Extract GloVe Embeddings


```python
import os
import urllib.request
import zipfile

glove_url = 'http://nlp.stanford.edu/data/glove.6B.zip'
glove_zip_path = 'glove.6B.zip'
extract_dir = 'glove.6B'

if not os.path.exists(extract_dir):
    print("Downloading GloVe embeddings... This might take a minute.")
    urllib.request.urlretrieve(glove_url, glove_zip_path)

    print("Extracting embeddings...")
    with zipfile.ZipFile(glove_zip_path, 'r') as zip_ref:
        zip_ref.extractall(extract_dir)
    print("Extraction complete!")
else:
    print("GloVe embeddings already downloaded and extracted.")
```

### Load Embeddings into Memory


```python
import numpy as np

def load_glove_embeddings(file_path):
    embeddings_index = {}
    print(f"Loading embeddings from {file_path}...")
    with open(file_path, 'r', encoding='utf-8') as f:
        for line in f:
            values = line.split()
            word = values[0]
            coefs = np.asarray(values[1:], dtype='float32')
            embeddings_index[word] = coefs
    print(f"Loaded {len(embeddings_index)} word vectors.")
    return embeddings_index

glove_path = os.path.join(extract_dir, 'glove.6B.100d.txt')
embeddings_dict = load_glove_embeddings(glove_path)
```

### Retrieve Vector Representations


```python
def get_word_vector(word, embeddings_dict):
    return embeddings_dict.get(word.lower(), None)

def get_sentence_vector(sentence, embeddings_dict, vector_dim=100):
    words = sentence.lower().split()
    word_vectors = [embeddings_dict[w] for w in words if w in embeddings_dict]
    if not word_vectors:
        return np.zeros(vector_dim)
    return np.mean(word_vectors, axis=0)


word = "science"
word_vector = get_word_vector(word, embeddings_dict)
if word_vector is not None:
    print(f"Vector representation for '{word}' (first 5 dimensions):", word_vector[:5])


sentence = "deep learning is fascinating"
sent_vector = get_sentence_vector(sentence, embeddings_dict)
print(f"Vector representation for '{sentence}' (first 5 dimensions):", sent_vector[:5])
```
