# Information Extraction (IE) from Structured/Unstructured Documents



### Common Tasks in IE:
1. **Named Entity Recognition (NER)**: Identifying names, organizations, locations, dates, etc.
2. **Relation Extraction**: Determining relationships between identified entities.
3. **Key-Value Pair Extraction**: Extracting specific fields from structured forms or invoices.
4. **Table Extraction**: Parsing tabular data from documents.

## Part 1: Named Entity Recognition (NER) on Unstructured Text



```python
import spacy
from spacy import displacy

# Load the small English pipeline from spaCy
nlp = spacy.load("en_core_web_sm")

# Sample unstructured text
sample_text = (
    "Google was founded on September 4, 1998, by Larry Page and Sergey Brin while "
    "they were Ph.D. students at Stanford University in California. "
    "Today, Alphabet Inc. serves as the parent company of Google and is headquartered in Mountain View."
)

# Process the text
doc = nlp(sample_text)

# Extract and print entities
print(f"Original Text:\n{sample_text}\n")
print("Extracted Entities:")
print(f"{'Entity':<25} | {'Label':<10} | {'Description'}")
print("-" * 70)
for ent in doc.ents:
    print(f"{ent.text:<25} | {ent.label_:<10} | {spacy.explain(ent.label_)}")
```

## Part 2: Extracting Structured Data from Semi-Structured Documents (HTML)



```python
from bs4 import BeautifulSoup
import pandas as pd

# Sample semi-structured HTML data
html_content = """
<html>
  <head><title>Company Profile: TechCorp</title></head>
  <body>
    <div class="metadata">
      <p><strong>Founded:</strong> 2010</p>
      <p><strong>Industry:</strong> Software and AI</p>
    </div>
    <h3>Key Executive Board</h3>
    <table id="executives">
      <thead>
        <tr>
          <th>Name</th>
          <th>Role</th>
        </tr>
      </thead>
      <tbody>
        <tr>
          <td>Alice Johnson</td>
          <td>CEO</td>
        </tr>
        <tr>
          <td>Bob Smith</td>
          <td>CTO</td>
        </tr>
      </tbody>
    </table>
  </body>
</html>
"""

soup = BeautifulSoup(html_content, "html.parser")

# 1. Extracting Key-Value Metadata
metadata = {}
meta_div = soup.find("div", class_="metadata")
if meta_div:
    for p in meta_div.find_all("p"):
        strong = p.find("strong")
        if strong:
          key = strong.text.replace(":", "").strip()
          value = p.text.replace(strong.text, "").strip()
          metadata[key] = value

print("Extracted Metadata (Key-Value Pairs):")
print(metadata)
print("\n" + "="*40 + "\n")

# 2. Extracting Tabular Data
table = soup.find("table", id="executives")
headers = [th.text for th in table.find_all("th")]
rows = []
for tr in table.find("tbody").find_all("tr"):
    rows.append([td.text for td in tr.find_all("td")])

df = pd.DataFrame(rows, columns=headers)
print("Extracted Table Data (DataFrame):")
print(df)
```
