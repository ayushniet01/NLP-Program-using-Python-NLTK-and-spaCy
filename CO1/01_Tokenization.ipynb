```python
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize

nltk.download('punkt')
nltk.download('punkt_tab')

story_text = "AI is becoming popular among students. It helps them solve complex problems efficiently."

sentence_tokens = sent_tokenize(story_text)
word_tokens = word_tokenize(story_text)

print("Sentences:")
print(sentence_tokens)

print("\nWords:")
print(word_tokens)
```

```python
import spacy

language_model = spacy.load("en_core_web_sm")

paragraph_text = "Robots are assisting humans in many industries. Their capabilities are improving rapidly."

processed_text = language_model(paragraph_text)

sentence_tokens_spacy = [part.text for part in processed_text.sents]
word_tokens_spacy = [part.text for part in processed_text]

print("Sentences:")
print(sentence_tokens_spacy)

print("\nWords:")
print(word_tokens_spacy)
```
