```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('wordnet')
nltk.download('omw-1.4')

from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.tokenize import word_tokenize
```

```python
sample_paragraph = "The researchers were studying various experiments while the students were writing detailed reports."

tokens_list = word_tokenize(sample_paragraph)

print("Original Words:")
print(tokens_list)
```

```python
stem_tool = PorterStemmer()

stemmed_words = [stem_tool.stem(item) for item in tokens_list]

print("Stemmed Words:")
print(stemmed_words)
```

```python
lemma_tool = WordNetLemmatizer()

lemmatized_words = [lemma_tool.lemmatize(item) for item in tokens_list]

print("Lemmatized Words:")
print(lemmatized_words)
```

```python
print("Word\t\tStem\t\tLemma")

for original_word, stem_value, lemma_value in zip(tokens_list, stemmed_words, lemmatized_words):
    print(f"{original_word}\t\t{stem_value}\t\t{lemma_value}")
```
