```python
import nltk
nltk.download('stopwords')
nltk.download('punkt')
nltk.download('punkt_tab')

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
document_text = "The students are working on their natural language processing project and they are learning many useful techniques."

document_words = word_tokenize(document_text)

print("Original Document:")
print(document_words)
```

```python
common_words = set(stopwords.words('english'))

filtered_document = [word for word in document_words if word.lower() not in common_words]

print("Document After Stop-word Removal:")
print(filtered_document)
```

```python
clean_text = " ".join(filtered_document)

print("Cleaned Text:")
print(clean_text)
```
