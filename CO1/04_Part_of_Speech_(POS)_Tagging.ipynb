```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')

from nltk.tokenize import word_tokenize
from nltk import pos_tag
```

```python
sentence_input = "The talented musician performed a beautiful melody during the evening concert."

sentence_tokens = word_tokenize(sentence_input)

print("Tokenized Sentence:")
print(sentence_tokens)
```

```python
tagged_sentence = pos_tag(sentence_tokens)

print("POS Tagged Sentence:")
print(tagged_sentence)
```

```python
for vocabulary_item, grammatical_tag in tagged_sentence:
    print(vocabulary_item, "->", grammatical_tag)
```
