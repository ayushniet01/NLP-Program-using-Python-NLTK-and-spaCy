```python
import nltk
import spacy

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')

from nltk.tokenize import word_tokenize
from nltk import pos_tag
from nltk.chunk import RegexpParser
```

```python
phrase_data = "The young scientist carefully examined the unusual chemical sample."

phrase_words = word_tokenize(phrase_data)
tagged_phrase = pos_tag(phrase_words)

print("POS Tagged Words:")
print(tagged_phrase)
```

```python
chunk_pattern = r"""
    NP: {<DT>?<JJ>*<NN.*>+}
    VP: {<RB>*<VB.*><NP|PP>*}
"""

regex_chunker = RegexpParser(chunk_pattern)
chunk_tree = regex_chunker.parse(tagged_phrase)

print("RegEx Chunking:")
print(chunk_tree)
```

```python
chunk_tree.pretty_print()
```

```python
spacy_model = spacy.load("en_core_web_sm")

parsed_sentence = spacy_model(phrase_data)

print("spaCy Dependency Parsing:")

for token_unit in parsed_sentence:
    print(token_unit.text, "->", token_unit.dep_, "->", token_unit.head.text)
```

```python
print("spaCy Noun Chunks:")

for noun_group in parsed_sentence.noun_chunks:
    print(noun_group.text)
```
