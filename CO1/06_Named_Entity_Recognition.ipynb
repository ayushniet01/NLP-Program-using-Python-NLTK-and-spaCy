```python
import spacy

ner_model = spacy.load("en_core_web_sm")
```

```python
news_text = "Elon Musk visited Berlin in March 2025 to discuss Tesla's expansion in Europe."

processed_news = ner_model(news_text)

print("Input Text:")
print(news_text)
```

```python
print("Named Entities:")

for entity_item in processed_news.ents:
    print(entity_item.text, "->", entity_item.label_)
```

```python
for entity_item in processed_news.ents:
    print("Entity:", entity_item.text)
    print("Label:", entity_item.label_)
    print("Description:", spacy.explain(entity_item.label_))
    print()
```
