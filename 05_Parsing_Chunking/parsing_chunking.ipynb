```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.chunk import RegexpParser

# Download required NLTK data
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
nltk.download('averaged_perceptron_tagger_eng', quiet=True)

# Sample sentence
sentence = "The quick brown fox jumps over the lazy dog."

# Tokenization
words = word_tokenize(sentence)

# POS Tagging
pos_tags = nltk.pos_tag(words)

# Noun Phrase Chunking
grammar = r"""
    NP: {<DT>?<JJ>*<NN.*>}
"""

chunk_parser = RegexpParser(grammar)

# Perform chunking
tree = chunk_parser.parse(pos_tags)

print("Words:")
print(words)

print("\nPOS Tags:")
print(pos_tags)

print("\nChunked Tree:")
print(tree)

print("\nParsing and Chunking setup completed!")
```

```python

import nltk
from nltk.tokenize import word_tokenize
from nltk.chunk import RegexpParser

# Download required data
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
nltk.download('averaged_perceptron_tagger_eng', quiet=True)

sentence = "Ayush is developing an intelligent chatbot using Python."

# ---------- REGEX CHUNKING ----------
print("========== REGEX CHUNKING ==========")

# Tokenization
words = word_tokenize(sentence)

# POS Tagging
pos_tags = nltk.pos_tag(words)

print("\nPOS Tagged Sentence:")
print(pos_tags)

# Noun Phrase Grammar
grammar = r"""
    NP: {<DT>?<JJ>*<NN.*>+}
"""

# Create parser
chunk_parser = RegexpParser(grammar)

# Parse sentence
chunk_tree = chunk_parser.parse(pos_tags)

print("\nChunking Result:")
chunk_tree.pretty_print()


# ---------- SIMPLE PARSING ----------
print("\n========== SIMPLE PARSING ==========")

print("\nWord\t\tPOS")
print("-" * 25)

for word, tag in pos_tags:
    print(f"{word:<15}{tag}")


# ---------- EXTRACT NOUN PHRASES ----------
print("\n========== NOUN PHRASES ==========")

for subtree in chunk_tree.subtrees():
    if subtree.label() == "NP":
        phrase = " ".join(word for word, tag in subtree.leaves())
        print("NP:", phrase)

print("\nParsing and Chunking completed!")
```
