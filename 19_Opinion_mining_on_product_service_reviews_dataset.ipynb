```python
import pandas as pd
from textblob import TextBlob

data = {
    'Review': [
        'This product is absolutely amazing! It exceeded all my expectations.',
        'Terrible service and cheap quality. I want a refund immediately.',
        'It is okay, does the job but has some minor issues.',
        'Highly recommended! Super fast delivery and great support.',
        'Worst experience ever. The app keeps crashing constantly.'
    ]
}

df = pd.DataFrame(data)

def analyze_sentiment(text):
    analysis = TextBlob(text)
    polarity = analysis.sentiment.polarity
    if polarity > 0:
        category = 'Positive'
    elif polarity < 0:
        category = 'Negative'
    else:
        category = 'Neutral'
    return pd.Series([polarity, analysis.sentiment.subjectivity, category])

df[['Polarity', 'Subjectivity', 'Sentiment']] = df['Review'].apply(analyze_sentiment)

display(df)
```
